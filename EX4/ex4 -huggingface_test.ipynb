```python
from transformers import pipeline
from PIL import Image
import os

#LOAD HUGGING FACE MODEL

classifier = pipeline(
    "image-classification",
    model="google/vit-base-patch16-224"
)

print("Hugging Face ViT model loaded successfully!")
```

```python
#SELECT SAMPLE IMAGE

image_path = "dataset/test/forest"

# Get first image from folder
image_files = os.listdir(image_path)

first_image = os.path.join(
    image_path,
    image_files[0]
)

print("\nTesting image:")
print(first_image)


#LOAD IMAGE
image = Image.open(first_image).convert("RGB")
```

```python
#PREDICT

results = classifier(image)

print("\nHugging Face Predictions:")
print("==========================")

for result in results[:5]:

    print(
        result["label"],
        "->",
        round(result["score"], 4)
    )
```
